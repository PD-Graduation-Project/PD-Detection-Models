# 1. Load dataloaders


```python
from data.dataloader import create_dataloaders

train_dataloader, val_dataloader = create_dataloaders(
    healthy_dir="../project_datasets/drawing/Healthy/",
    pd_dir="../project_datasets/drawing/Parkinson/",

    img_size=(256, 256),
    batch_size=32,
)
```

```python
train_inception, val_inception = create_dataloaders(
    healthy_dir="../project_datasets/drawing/Healthy/",
    pd_dir="../project_datasets/drawing/Parkinson/",

    img_size=(299, 299),  # minimum input size for inceptionV3 model
    batch_size=32,
)
```

# 2. Load models

- **DenseNet161** – 28.7M parameters
- **EfficientNet-B2** – 9.2M parameters
- **InceptionV3** – 23.8M parameters
- **MobileNetV3-Large** – 5.4M parameters
- **ResNet34** – 21.8M parameters


```python
from Models import (
    densenet161,
    efficientnetB2,
    InceptionV3,
    mobilenetV3,
    resnet34
)

models = [
    densenet161.DenseNet161Binary(),
    efficientnetB2.EfficientNetB2Binary(),
    InceptionV3.InceptionV3Binary(),
    mobilenetV3.MobileNetV3LargeBinary(),
    resnet34.ResNet34Binary()
]

model_names = [
    "DenseNet161",
    "EfficientNet-B2",
    "InceptionV3",
    "MobileNetV3-Large",
    "ResNet34"
]
```

# 3. Train models


```python
from training.trainer import train

for model, model_name in zip(models, model_names):
    train(
        model=model,
        train_dataloader=train_inception if model_name == "InceptionV3" else train_dataloader,
        val_dataloader=val_inception if model_name == "InceptionV3" else val_dataloader,

        model_name=model_name,
        run_name=model_name,

        epochs=20
    )
```
