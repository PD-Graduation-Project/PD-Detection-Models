# 1. Load dataloaders 
> With `CLAHE`

```python
from data.dataloader import create_dataloaders

train_dataloader, val_dataloader = create_dataloaders(
    healthy_dir= "../project_datasets/drawing/Healthy/",
    pd_dir= "../project_datasets/drawing/Parkinson/",
    
    img_size=(224, 224),
    batch_size= 32,
)
```

# 2. Load model

- DenseNet201 (20.0M)

```python
from Models.improved_DenseNet201 import create_improved_densenet

model = create_improved_densenet()

model_name = "DenseNet201"
```

# 3. Train Model

```python
from training.trainer import train

train(
    model= model,
    train_dataloader= train_dataloader,
    val_dataloader= val_dataloader,
    
    model_name= model_name,
    run_name= "DenseNet201_initial",
    
    epochs= 15
)
```
