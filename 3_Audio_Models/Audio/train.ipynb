# 1. Load dataloaders


```python
from data.dataloader import create_dataloaders

train_dataloader, val_dataloader = create_dataloaders(
    healthy_dir="../../project_datasets/audio/audio/Healthy/",
    pd_dir="../../project_datasets/audio/audio/Parkinson/",

    batch_size=32,
)
```

# 2. Load models

- **Wav2Vec2 Base** – ~95M parameters
- **HuBERT Base** – ~95M parameters  
- **WavLM Base** – ~95M parameters


```python
from Models import (
    HuBERTBinary,
    Wav2Vec2Binary,
    WavLMBinary
)

models = [
    HuBERTBinary.HuBERTBinary(),
    Wav2Vec2Binary.Wav2Vec2Binary(),
    WavLMBinary.WavLMBinary(),
]

model_names = [
    "HuBERT",
    "Wav2Vec2",
    "WavLM",
]
```

# 3. Train models


```python
from training.trainer import train

for model, model_name in zip(models, model_names):
    train(
        model=model,
        train_dataloader=train_dataloader,
        val_dataloader=val_dataloader,

        model_name=model_name,
        run_name=model_name,

        epochs=20
    )
```
